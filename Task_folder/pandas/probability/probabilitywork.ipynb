# Probability

```python
orders_reviews.head(2)
```

```python
#probability of 5- star review 
# out of all orders,what is the probability that a review has 5 star rating
# this tells us how lkely a customer is to give the maximum satisfaction score 

# p(5star)= number of 5 - stars review/total reviews 

total_reviews = len(orders_reviews)
favourable = (orders_reviews['review_score']==5).sum()
p_5star = favourable/total_reviews

print(f'Probability of 5 stars = {round(p_5star * 100,2)} %')
```

**Interpretation of result** 

Approximately 58% of customers give 5 stars.this is the baseline success probability of excellent reviews.

2. Complementary Probability

Concept: P(A Complement) = 1 -P(A)
Question: what is the probability that a review is not 5 star 

```python
# if we know the probability of 5 stars, what is the probability of not getting a 5 star rating?

# this means any rating from 1-4
# we are still going to se the review dataset
# formula style 
p_not5 = (order_reviews['review_score'] != 5).sum()/len(order_reviews)
print(f"P(Review ! = 5 star) = {round(p_not5*100,2)}%")
```

Interpretation

Approximately 42% of reviews are less than perfect.This highlights that not everyone is fully satisfied with the product purchased 

3. Joint Probability

Concept P(AnB)

Question: What is the probability that an order was paid by credit  and received a 5 - star review?

```python
# this gives the likelihood of two events happening together.
# here we will be using the orders_payments and orders_reviews. and we have merged them already 
# lets merge orders_Payments  and order_reviews 
order_reviews.head(2)

orders_payment_reviews = pd.merge(order_reviews,orders_payments,on='order_id',how='left')
orders_payment_reviews.head(2)
```

```python
orders_payment_reviews.columns
```

```python
total_orders = len(orders_payment_reviews)
favourable_joint =((orders_payment_reviews['payment_type']=='credit_card') & (orders_payment_reviews['review_score']==5)).sum() 

p_joint = favourable_joint/total_orders

print(f"P(Credit Card ∩ 5 stars) = {round(p_joint *100, 2)}%")
total_orders
```

Interpretation 

Approximately 38% of all orders are bth paid by credit card and get a perfect review 


# 4. Conditional Probability 

Concept:

`𝑃(𝐴∣𝐵)=𝑃(𝐴∩𝐵)/𝑃(𝐵)` 
                
`𝑃(𝐴∣𝐵)= Favourable_outcomes / Total_credit_card_outcomes`

Question: What is the probability that a reviw is 5 star given tht the payment was made by credit card?

```python
orders_payment_reviews.head(2)
```

```python
favourable_cond = ((orders_payment_reviews['payment_type']=='credit_card') & (orders_payment_reviews['review_score']==5)).sum()
total_credit =(orders_payment_reviews['payment_type']=='credit_card').sum()

p_cond = favourable_cond/total_credit

print(f"P(5 stars | Credit Card) = {round(p_cond *100, 2)}%")


# for  P(5 stars | voucher)
favourable_cond = ((orders_payment_reviews['payment_type']=='voucher') & (orders_payment_reviews['review_score']==5)).sum()
total_credit = (orders_payment_reviews['payment_type']=='voucher').sum()
p_cond = favourable_cond/total_credit
print(f"P(5 stars | voucher) = {round(p_cond *100, 2)}%")


favourable_cond = ((orders_payment_reviews["payment_type"]=="boleto") & (orders_payment_reviews["review_score"]==5)).sum()

total_credit = (orders_payment_reviews["payment_type"]=="boleto").sum()

p_cond = favourable_cond / total_credit
print(f"P(5 stars | boleto) = {round(p_cond *100, 2)}%")
```

# Interpretation

Each one is around 55-57% ,meaning in every payment method  about half of customers gave 5  star rating 



# 5.  Bayes's Rule 

Concept:
P(A|B) = P(B|A).P(A)/P(B)

A=payment

B =Review=5

`P(payment | review=5) = P(review=5 | payment).p(payment) / p(review =5)`

Question: Given that an order received a 5-star review, what is the probability that it was paid with each payment method (credit_card, boleto, voucher)?

Among only the 5-star reviews, how are payment methods distributed?



```python
total_5star = (orders_payment_reviews['review_score']==5).sum()

# P(payment = creditcard | review = 5 ) --- 
favourable_cond = ((orders_payment_reviews['payment_type']=='credit_card') & (orders_payment_reviews['review_score']==5)).sum()

p_cond = favourable_cond/total_5star
print(f"P(Credit Card | 5 stars) = {round(p_cond * 100, 2)}%")


# for P(payment = voucher | review = 5)
favourable_cond = ((orders_payment_reviews["payment_type"]=="voucher") & (orders_payment_reviews["review_score"]==5)).sum()

p_cond = favourable_cond / total_5star
print(f"P(Voucher | 5 stars) = {round(p_cond * 100, 2)}%")

#P(payment = boleto | review = 5)
favourable_cond = ((orders_payment_reviews["payment_type"]=="boleto") & (orders_payment_reviews["review_score"]==5)).sum()

p_cond = favourable_cond / total_5star
print(f"P(Boleto | 5 stars) = {round(p_cond * 100, 2)} %")

```

# Interpretation 

1. P(Credit card| 5 stars) = among all 5 star reviews,what fractions used credits card? Among 5 - star reviews, 65.74% used credit card 

2. P(Voucher| 5 stars)= Among all 5-star reviews, what fractions used voucher? Among 5-star review,5.39% used boleto

3. P(Boleto|5 stars) = Among all 5-star reviews, what fraction used boleto? 18.95% of 5-star reviws uses baleto payment method 

# Expectation and Variance

**Question1**

What is the average delivery time (expectation) and variance (spread/consistency) of deliveries?

```python
orders_payment_reviews.head(2)
```

```python
orders_payment_reviews['order_purchase_timestamp']= pd.to_datetime(orders_payment_reviews['order_purchase_timestamp'])
orders_payment_reviews['order_delivered_customer_date'] = pd.to_datetime(orders_payment_reviews['order_delivered_customer_date'])

#calculate delivery time in days 
orders_payment_reviews['delivery_time_days'] = (orders_payment_reviews['order_delivered_customer_date'] - orders_payment_reviews['order_purchase_timestamp']).dt.days

# drop missing value
delivery_times =orders_payment_reviews['delivery_time_days'].dropna()

# lets get the expectation(mean)

mean_delivery = delivery_times.mean()

# Variance & Standard Deviation
variance_delivery = delivery_times.var()

std_delivery = delivery_times.std()

print("Expected delivery time (Mean):", round(mean_delivery, 2), "days")
print("Variance of delivery time:", round(variance_delivery, 2))
print("Standard Deviation of delivery time:", round(std_delivery, 2), "days")
```

```python
# Expectation (Mean review score)
mean_review = orders_payment_reviews["review_score"].mean()

# Variance & Standard Deviation
variance_review = orders_payment_reviews["review_score"].var()
std_review = orders_payment_reviews["review_score"].std()

print("Expected review score (Mean):", round(mean_review, 2))
print("Variance of review score:", round(variance_review, 2))
print("Standard Deviation of review score:", round(std_review, 2))
```

The mean review score shows the overall satisfaction (e.g ~4 stars = mostly happy customers). The std tells us that most customers give between 3-5 stars . if S.D was very high (e.g 2), it would mean customers are divided (some give 1, others give 5). But here, the service reputation is fairly consistent
